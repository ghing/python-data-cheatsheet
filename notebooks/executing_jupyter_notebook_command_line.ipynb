# Executing a Jupyter notebook from the command line

## Using `jupyter run`

In a somewhat recent feature addition, there's now a subcommand that can be used to run notebooks from the command line:

```
jupyter run notebook.ipynb
```

Unlike using `nbconvert`, this doesn't create copies of the notebook as a result of the execution.

Source: [Using a command-line interface](https://docs.jupyter.org/en/latest/running.html#using-a-command-line-interface)

## Using `nbconvert`

Before the advent of `jupyer run`, the way to execute a notebook was to use `nbconvert`:

```
jupyter nbconvert --to notebook --execute mynotebook.ipynb
```

This creates a similarly-named copy of the notebook. Since I usually want an HTML version, similar to how RStudio automatically creates an HTML version of RMarkdown notebooks, I usually use a somewhat different set of parameters:

```
jupyter nbconvert \
	  --to html \
	  --output notebook.nb.html \
	  --execute \
	  notebook.ipynb
```

Source: [Executing notebooks from the command line](https://nbconvert.readthedocs.io/en/latest/execute_api.html#executing-notebooks-from-the-command-line)

## Using papermill

I've never used this, but the `jupyter run` documentation recommends using [papermill](https://github.com/nteract/papermill) for more advanced execution needs, including passing command-line parameters to the notebook.
